#tuple
##ordered collection of items()
##one main difference with the tuple is immutable
# var_name = list[[stu1:()],[stu2:()]..]

```python
#create a tuple
student =(101,"sanjeev","cse",8.9)
print(student)

#tuples using index
print("student info in first location" ,student[0])
print(student[1])
#negative index
print("hiiiii", student[-2])
print(student[-1])
```

```python
# to find the length of tuple
subjects=("dsa","dsv","os","python","ml")
print(len(subjects))
```

```python
#slicing
subjects=("dsa","dsv","os","python","ml")
print(subjects[:3])
print(subjects[2:5])
print(subjects[4:])
```

```python
#tuple immutable
students_info=("Chirag",21,"cse",8.7)
students_info[3]=9.1
print(students_info)
```

```python
#nested tuple
student=(101,"rahul",(89,78,66))
print(student)
print(student[1])
print(student[2])
print(student[2][1])
```

```python
# count function
marks=(75,78,68,75,89,75,92,80)
print("frequency",marks.count(75))
```

```python
#find the elements present in tuiple or not
subjects=("python","dsv","sql","ml","stats")
print("python" in subjects)
print("java" in subjects)
```

#SET
## Unordered collections of items
## mutable
## allows the different data types
## it doesn't support for index
## it gives output with unique values(NO duplication)

```python
# symbol for set { }
# create a set with similar data items
marks={75,89,75,68,93,55,83,75,67,84}
print(marks)
```

```python
# create a empty set
a={}
a=set()
#check its type
print(type(a))
```

```python
# set doesnt support the index
students={"amit","riya" ,"priya"}
print(students[1])
```

```python
# add an item in the set
students={"amit","riya" ,"priya"}
students.add("rahul")
print(students)
students.update(["ram","sam","hari"])
print(students)
# remove item from the set
students.remove("ram")
print(students)
students.discard("sam")
print(students)
students.discard("karan")
print(students)
# clear list of items
students.clear()
print(students)
```

```python
marks={99.88,88,99}
del marks
print(marks)
```

```python
# set operations
students={"amit","neha","sam","karan"}
stud={"amit","danish","sam","heera","neha"}
#for union operation using set
result=students.union(stud)
print(result)
#for intersection operation using set
inter=students.intersection(stud)
print(inter)
#for difference operation using set
diff=students.difference(stud)
print(diff)
dif=stud.difference(students)
print(dif)
```

```python
# convert list to set and vice-versa
id=[101,102,103,104,105,106,102,103,102,103,105]
stud=set(id)
print(stud)
roll=list(stud)
print(roll)
```
