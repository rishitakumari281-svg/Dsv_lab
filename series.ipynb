##Python Pandas Data Structure
the pandas provides two data structures for processing the data:
####Series
####Data frame
#series
it is defined as 1 dimensional array that is capable of store various data type.


```python
import pandas as pd
import numpy as np
info =np.array(['P','a','n','d','a','s'])
a=pd.Series(info)
print(a)
```

```python
#dictionary
import pandas as pd
import numpy as np
d={
    "Name": np.array(["Tarun","Janvi","Samit","Rishi"]),
    "Course": np.array(["CSE","ECE","BIOTECH","BCA"]),
    "Age": np.array([20,21,20,21])
}
print(d)
df=pd.DataFrame(d)
print(df)
```

```python
#question 29
import pandas as pd

sales = pd.Series([300, 800, 1200, 450], index=['ProductA','ProductB','ProductC','ProductD'])

print("Values:", sales.values)
print("Index:", sales.index)

print("\nSales of ProductC:", sales['ProductC'])

print("\nSales > 500:\n", sales[sales > 500])

print("\nIs 'ProductB' in sales?", 'ProductB' in sales)

sales_dict = sales.to_dict()
print("\nDictionary:", sales_dict)
```

```python
#question 30
import pandas as pd

data = {'A': 15, 'B': 35, 'C': 25}
s = pd.Series(data, index=['C','A','D','B'])

s.name = 'Observations'
s.index.name = 'Category'

print(s)
print("\nFilter > 20:\n", s[s > 20])
print("\nNotnull:\n", pd.notnull(s))
```
