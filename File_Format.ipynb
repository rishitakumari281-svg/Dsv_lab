```python
CSE=pd.DataFrame({
    "Name" :["Amit","Riya"],
    "Marks" :[89,87,]
})
ECE=pd.DataFrame({
    "Name" :["Rahul","Priya"],
    "Marks" :[96,78]
})
IT=pd.DataFrame({
    "Name" :["Rohit","Ram"],
    "Marks" :[65,98]
})
with pd.ExcelWriter("department.xlsx") as writer:
  CSE.to_excel(writer,sheet_name="CSE",index=False)
  ECE.to_excel(writer,sheet_name="ECE",index=False)
  IT.to_excel(writer,sheet_name="IT",index=False)
```
