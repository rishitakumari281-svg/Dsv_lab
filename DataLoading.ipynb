```python

import pandas as pd
import os

# Create DataFrame
data = {
    "Name": ["Amit", "Riya", "Rahul", "Priya", "Karan"],
    "Age": [20, 21, 20, 22, 21],
    "Department": ["CSE", "IT", "CSE", "ECE", "IT"],
    "Marks": [85, 90, 78, 88, 76]
}

df = pd.DataFrame(data)

# Include 5 more rows in the DataFrame
new_data = {
    "Name": ["Neha", "Arjun", "Sneha", "Rohit", "Ananya"],
    "Age": [20, 22, 21, 20, 23],
    "Department": ["ECE", "CSE", "IT", "CSE", "ECE"],
    "Marks": [92, 81, 87, 79, 95]
}

df_new = pd.DataFrame(new_data)
df = pd.concat([df, df_new], ignore_index=True)

# Inspect the DataFrame
print("Shape:\n", df.shape)
print("Head with 3 rows:\n", df.head(3))
print("Tail of 2 rows:\n", df.tail(2))
print("Index:\n", df.index)
print("Columns:\n", df.columns.tolist())
print("Values:\n", df.values)
print("Data types:\n", df.dtypes)
print("Statistics:\n", df.describe())
print("Information:")
df.info()

# Convert into different file formats
df.to_csv("students.csv", index=False)
df.to_excel("students.xlsx", index=False)
df.to_json("students.json", orient="records", indent=4)
df.to_parquet("students.parquet", index=False)

# Compare file sizes
files = ["students.csv", "students.xlsx", "students.json", "students.parquet"]

print("\nFile Size Comparison:")
for file in files:
    size = os.path.getsize(file)
    print(file, ":", size, "bytes")
```
