# Q3.1 — Difference Between is and ==

# == checks whether two objects have the same value.
# is checks whether two variables refer to the exact same object in memory.

```python
# Q3.2 — Equality vs Identity

a = [1, 2, 3]
b = [1, 2, 3]

print(a == b)
print(a is b)

# Output:
# True
# False
```

```python
# Q3.3 — Comparing Small and Large Integers

x = 100
y = 100

print(x is y)
print(x == y)

p = 1000
q = 1000

print(p is q)
print(p == q)

# Small integers are often reused by Python,
# while larger integers may be stored as different objects.
```

```python
# Q7.1 — Creating an Absolute Value Function

def my_abs(val):
    if val < 0:
        return 0 - val
    return val

print(my_abs(7))
print(my_abs(-7))
print(my_abs(0))
```

# Q7.2 — Passing a String to my_abs()

# my_abs("Hi") will produce a TypeError
# because a string cannot be compared with an integer using <.

```python
# Q7.3 — Printing Instead of Returning an Absolute Value

def print_abs(val):
    if val < 0:
        print(0 - val)
    else:
        print(val)

x = print_abs(-2.7)

print(x)

# Output:
# 2.7
# None
```

```python
# Q7.4 — Local Variable Inside a Function

def inc_val(val):
    val = val + 1

x = 7

inc_val(x)

print(x)

# Output:
# 7
```

```python
# Q7.5 — Swapping Local Variables

def swap_message(val1, val2):
    temp = val1
    val1 = val2
    val2 = temp

    print("Inside function:", val1, val2)


x = 6
y = 3

swap_message(x, y)

print("Original variables:", x, y)
```

```python
# Q8.1 — Understanding Local Scope

def my_abs(val):
    if val < 0:
        return 0 - val
    return val

print(val)

# This produces a NameError because val only exists
# inside the my_abs() function.
```

```python
# Q8.2 — Reading a Global Variable

counter = 0

def show_counter():
    print(counter)

show_counter()
```

```python
# Q8.3 — Modifying a Global Variable Without global

counter = 0

def increment_counter():
    counter = counter + 1

increment_counter()

# This produces an UnboundLocalError because Python
# treats counter as a local variable inside the function.
```

```python
# Q8.4 — Modifying a Global Variable Using global

counter = 0

def increment_counter():
    global counter
    counter = counter + 1

increment_counter()

print(counter)
```
